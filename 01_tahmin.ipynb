```python
import requests, matplotlib.pyplot as plt

params = {"latitude": 41.0, "longitude": 39.7,      # kendi koordinatını yaz
          "hourly": "temperature_2m,precipitation",
          "forecast_days": 3, "timezone": "auto"}
d = requests.get("https://api.open-meteo.com/v1/forecast", params=params).json()["hourly"]

fig, ax1 = plt.subplots(figsize=(10, 4))
ax1.plot(d["time"], d["temperature_2m"], color="red")
ax1.set_ylabel("Sıcaklık (°C)")
ax2 = ax1.twinx()
ax2.bar(d["time"], d["precipitation"], alpha=0.4)
ax2.set_ylabel("Yağış (mm)")
ax1.set_xticks(d["time"][::12]); ax1.tick_params(axis="x", rotation=45)
plt.tight_layout(); plt.show()
```
