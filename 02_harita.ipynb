```python
import requests, numpy as np, matplotlib.pyplot as plt
import cartopy.crs as ccrs, cartopy.feature as cf

lats, lons = np.arange(36, 43, 1.0), np.arange(26, 46, 1.0)
LA, LO = np.meshgrid(lats, lons, indexing="ij")
params = {"latitude": ",".join(map(str, LA.ravel())),
          "longitude": ",".join(map(str, LO.ravel())),
          "current": "temperature_2m,wind_speed_10m,wind_direction_10m"}
res = requests.get("https://api.open-meteo.com/v1/forecast", params=params).json()

T  = np.array([r["current"]["temperature_2m"] for r in res]).reshape(LA.shape)
ws = np.array([r["current"]["wind_speed_10m"] for r in res]).reshape(LA.shape)
wd = np.radians([r["current"]["wind_direction_10m"] for r in res]).reshape(LA.shape)
u, v = -ws*np.sin(wd), -ws*np.cos(wd)

ax = plt.figure(figsize=(11, 5)).add_subplot(projection=ccrs.PlateCarree())
ax.set_extent([25.5, 45.5, 35.5, 42.5])
cs = ax.contourf(lons, lats, T, levels=20, cmap="coolwarm")
ax.quiver(lons, lats, u, v)
ax.coastlines(); ax.add_feature(cf.BORDERS, linestyle=":")
plt.colorbar(cs, label="°C"); plt.title("Şu anki sıcaklık ve rüzgâr"); plt.show()
```
